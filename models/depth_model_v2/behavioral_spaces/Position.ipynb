```python
import os
os.environ['AWS_PROFILE'] = 'admin'
os.environ['HAVEN_DATABASE'] = 'haven'

import numpy as np
import pandas as pd
from scipy.stats import norm
from random import sample
import h3

from mirrorverse.utils import read_data_w_cache
from haven.db import write_data
```

```python
sql = '''
select distinct 
    h3_index 
from 
    copernicus_physics
'''
data = read_data_w_cache(sql)
print(data.shape)
data.head()
```

```python
data['lat'] = round(data['h3_index'].apply(lambda h: h3.h3_to_geo(h)[0])) / 90
data['lon'] = round(data['h3_index'].apply(lambda h: h3.h3_to_geo(h)[1])) / 180
data['version'] = 1
data.head()
```

```python
write_data(data, 'h3_to_position', ['version'])
```
